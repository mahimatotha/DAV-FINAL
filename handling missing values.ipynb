```python
import pandas as pd
import numpy as np
data={
    "student":["A","B","C","D","E"],
    "gender":["male","female","male","female","male"],
    "studyhours":[5,6,np.nan,4,7],
    "attendence":[80,85,75,np.nan,90],
    "marks":[75,82,68,55,np.nan],
    "result":["pass","pass","pass","fail","pass"]
}
df=pd.DataFrame(data)
print(df)
    
```

```python
print("\nmissing values\n")
print(df.isnull().sum())
```

```python
#mean imputation
df["studyhours"]=df["studyhours"].fillna(df["studyhours"].mean())
#just checking wherther the missing values is replaced or not
print("\nmissing values\n")
print(df.isnull().sum())
```

```python
#median imputation
df["marks"]=df["marks"].fillna(df["marks"].median())
print("\nmissing values\n")
print(df.isnull().sum())
print(df)

```

```python
df["attendence"]=df["attendence"].ffill()
print(df)
```

```python
df["gender"]=df["gender"].map({"male":1,"female":0})
print(df)
```

```python
df["result"]=df["result"].map({"pass":1,"fail":0})
print(df)
```
