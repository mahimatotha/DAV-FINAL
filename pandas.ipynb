```python
import numpy as np
```

```python
arr=np.array([10,20,30,40,50])
```

```python
print("numpy array:")
print(arr)
```

```python
print("array indexing")
print("first element:",arr[0])
print("third element:",arr[2])
print("last element:",arr[-1])
```

```python
print("array slicing")
print("first three elements:",arr[0:3])
print("elements from 3rd position",arr[2:])
print("last three elements",arr[-3:])
```

```python
print("array operations")
print("addition: ",arr+5)
print("subtraction: ",arr-5)
print("multiplication: ",arr*2)
print("division: ",arr/10)
```

```python
print("mathematical operations")
print("sum:",np.sum(arr))
print("mean:",np.mean(arr))
print("maximum:",np.max(arr))
print("minimum: ",np.min(arr))
```

```python
print("-------array reshaping-------")
reshaped_array=arr.reshape(5,1)
print("reshaped array is: ")
print(reshaped_array)
```

```python
import pandas as pd
```

```python
data={
    "Name":["asha","ravi","meera","karan"],
    "marks":[88,72,91,65],
    "branch":["cse","ece","cse","ise"]
}
df=pd.DataFrame(data)
print("pandas dataframe is: ")
print(df)
```

```python
print("column uindexing")
print("name column is:")
print(df["Name"])
```

```python
print("marks colum is: ")
print(df["marks"])
```

```python
print("row indexing")
print("first row")
print(df.iloc[0])
print("\n")
print(df.iloc[1])
```

```python
print("specific value")
print("marks of first student:",df.iloc[0,1])
```

```python
print("row slicing")
print("first two rows")
print(df.iloc[0:2])
```

```python
print("row and column slicing")
print("first 3 rows with name and marks columns")
print(df.iloc[0:3, [0,1]])
```

```python
print("data filtering")
result=df[df["marks"]>70]
print("students with marks greater than 70 is: ")
print(result)
```
